## 1. InfoNCE Loss — The Math Made Visual

Given a batch of N image-text pairs:

$$\mathcal{L}_{\text{InfoNCE}} = -\frac{1}{N} \sum_{i=1}^{N} \log \frac{\exp(\text{sim}(I_i, T_i) / \tau)}{\sum_{j=1}^{N} \exp(\text{sim}(I_i, T_j) / \tau)}$$

Where:
- $\text{sim}(I_i, T_i)$ = cosine similarity of matching pair (positive)
- $\text{sim}(I_i, T_j)$ for $j \neq i$ = similarity with non-matching (negatives)  
- $\tau$ = temperature (controls sharpness)

## Mutual Information Lower Bound

InfoNCE is not just a convenient loss function — it is a **provable lower bound on mutual information** between two views (e.g., image and text representations).

### The Bound

$$I(X; Y) \geq \log(N) - \mathcal{L}_{\text{InfoNCE}}$$

where:
- $I(X; Y)$ — mutual information between image embeddings $X$ and text embeddings $Y$
- $N$ — batch size (number of negative samples)
- $\mathcal{L}_{\text{InfoNCE}}$ — the InfoNCE loss value

**Minimizing InfoNCE loss maximizes a lower bound on mutual information** between aligned image-text pairs.

### Batch Size and the MI Bound

The $\log(N)$ term shows that **larger batch sizes yield tighter bounds** on mutual information — more negatives → better estimate of the true MI.

| Batch Size $N$ | $\log_2(N)$ (bits) | MI Lower Bound (if $\mathcal{L} = 0$) |
|----------------|---------------------|---------------------------------------|
| 32 | 5.0 | 3.5 bits (typical $\mathcal{L} \approx 1.5$) |
| 256 | 8.0 | 5.5 bits (typical $\mathcal{L} \approx 2.5$) |
| 32,768 | 15.0 | 10.3 bits (typical $\mathcal{L} \approx 4.7$) |

CLIP's batch size of **32,768** achieves an MI bound of approximately **10.3 bits** — a remarkably tight estimate that explains much of CLIP's representation quality.

### Practical Implication

This theoretical connection explains two empirical observations:
1. **Larger batches consistently improve contrastive learning** — not just because of more negatives, but because the MI bound becomes tighter
2. **The loss cannot go below $\log(N) - I(X;Y)$** — there is a fundamental floor determined by batch size and true mutual information

## Gradient Analysis of InfoNCE

Understanding the gradients reveals **why hard negatives matter** and how InfoNCE actually learns representations.

For a single row $i$ in the image→text direction, the loss is:

$$\mathcal{L}_i = -\log p_{ii}, \quad p_{ij} = \frac{\exp(s_{ij}/\tau)}{\sum_k \exp(s_{ik}/\tau)}$$

Taking the gradient with respect to the logit $s_{ij}$:

$$\frac{\partial \mathcal{L}_i}{\partial s_{ij}} = \frac{1}{\tau}\left(p_{ij} - \mathbb{1}[j = i]\right)$$

### Gradient Behavior

| Pair Type | Gradient w.r.t. $s_{ij}$ | Direction | Interpretation |
|-----------|---------------------------|-----------|----------------|
| **Positive** $(i, i)$ | $p_{ii} - 1$ (always **negative**) | ↑ similarity | Pull matching pair together |
| **Negative** $(i, j), j \neq i$ | $p_{ij}$ (always **positive**) | ↓ similarity | Push non-matching pair apart |

The magnitude of each gradient equals the **current softmax probability** $p_{ij}$.

### Why Hard Negatives Drive Learning

| $p_{ij}$ (neg prob) | Gradient | Push Strength |
|---------------------|----------|---------------|
| 0.90 | 0.90 | **Strong** — model confuses this negative with the anchor |
| 0.50 | 0.50 | Moderate |
| 0.10 | 0.10 | Weak |
| 0.01 | 0.01 | **Negligible** — model already separates this pair |

**Key insight:** Easy negatives ($p_{ij} \approx 0$) contribute almost zero gradient. Hard negatives ($p_{ij}$ high) receive strong push-apart signals. This is why larger batches and hard negative mining are critical — they ensure the model always encounters negatives that produce meaningful learning signals.

```python
def infonce_loss_detailed(img_emb, txt_emb, temperature=0.07):
    """InfoNCE with step-by-step computation for visualization."""
    # Step 1: Compute cosine similarity
    img_norm = F.normalize(img_emb, dim=-1)
    txt_norm = F.normalize(txt_emb, dim=-1)
    sim_matrix = img_norm @ txt_norm.T  # [N, N]
    
    # Step 2: Scale by temperature
    logits = sim_matrix / temperature
    
    # Step 3: Labels = diagonal (matching pairs)
    N = img_emb.shape[0]
    labels = torch.arange(N)
    
    # Step 4: Cross-entropy loss (both directions)
    loss_i2t = F.cross_entropy(logits, labels)
    loss_t2i = F.cross_entropy(logits.T, labels)
    loss = (loss_i2t + loss_t2i) / 2
    
    # Step 5: Probabilities for visualization
    probs_i2t = F.softmax(logits, dim=1)
    
    return {
        'loss': loss,
        'sim_matrix': sim_matrix.detach(),
        'logits': logits.detach(),
        'probs': probs_i2t.detach(),
        'labels': labels
    }


# Visualize each step
N = 5
img_emb = torch.randn(N, 64)
txt_emb = torch.randn(N, 64)

result = infonce_loss_detailed(img_emb, txt_emb, temperature=0.07)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
fig.suptitle('InfoNCE Loss — Step by Step', fontsize=16, fontweight='bold')

# Step 1: Raw similarity
ax = axes[0]
sim = result['sim_matrix'].numpy()
im = ax.imshow(sim, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_title('Step 1: Cosine Sim')
for i in range(N):
    for j in range(N):
        ax.text(j, i, f'{sim[i,j]:.2f}', ha='center', va='center', fontsize=9)
plt.colorbar(im, ax=ax, shrink=0.8)

# Step 2: Scaled logits
ax = axes[1]
logits = result['logits'].numpy()
im = ax.imshow(logits, cmap='RdBu_r')
ax.set_title('Step 2: Logits (sim/τ)')
for i in range(N):
    for j in range(N):
        ax.text(j, i, f'{logits[i,j]:.1f}', ha='center', va='center', fontsize=9)
plt.colorbar(im, ax=ax, shrink=0.8)

# Step 3: Target
ax = axes[2]
target = np.eye(N)
ax.imshow(target, cmap='Greens')
ax.set_title('Step 3: Target Labels')
for i in range(N):
    for j in range(N):
        ax.text(j, i, '1' if i==j else '0', ha='center', va='center', fontsize=12)

# Step 4: Softmax probabilities
ax = axes[3]
probs = result['probs'].numpy()
im = ax.imshow(probs, cmap='YlOrRd', vmin=0, vmax=1)
ax.set_title(f'Step 4: Softmax Probs\nLoss = {result["loss"].item():.3f}')
for i in range(N):
    for j in range(N):
        ax.text(j, i, f'{probs[i,j]:.2f}', ha='center', va='center', fontsize=9)
plt.colorbar(im, ax=ax, shrink=0.8)

for ax in axes:
    ax.set_xlabel('Text')
    ax.set_ylabel('Image')

plt.tight_layout()
plt.savefig('../assets/infonce_steps.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"\nGoal: Make diagonal probabilities → 1.0, off-diagonal → 0.0")
```

## 2. Temperature — The Most Important Hyperparameter

```python
# Effect of temperature on the similarity distribution

temperatures = [0.01, 0.07, 0.5, 1.0, 5.0]
sim_scores = torch.tensor([0.8, 0.3, 0.1, -0.2, -0.5])  # similarity with 5 items

fig, axes = plt.subplots(1, len(temperatures), figsize=(20, 4))
fig.suptitle('Effect of Temperature (τ) on Probability Distribution', 
             fontsize=14, fontweight='bold')

for ax, tau in zip(axes, temperatures):
    probs = F.softmax(sim_scores / tau, dim=0).numpy()
    colors = ['#2ECC71'] + ['#E74C3C'] * 4  # positive = green
    ax.bar(range(5), probs, color=colors, alpha=0.8)
    ax.set_title(f'τ = {tau}', fontsize=12, fontweight='bold')
    ax.set_ylim(0, 1.05)
    ax.set_xticks(range(5))
    ax.set_xticklabels(['pos', 'neg1', 'neg2', 'neg3', 'neg4'], fontsize=8)
    
    # Annotate
    for i, p in enumerate(probs):
        ax.text(i, p + 0.02, f'{p:.3f}', ha='center', fontsize=8)

plt.tight_layout()
plt.savefig('../assets/temperature_effect.png', dpi=150, bbox_inches='tight')
plt.show()

print("Low τ (0.01) → Very sharp, picks the highest similarity")
print("High τ (5.0) → Very uniform, all items look similar")
print("Sweet spot (0.07) → Discriminative but not too sharp")
```

## Learnable Temperature in CLIP

Rather than fixing temperature as a hyperparameter, **CLIP learns it during training** by parameterizing:

$$\tau = \exp(t), \quad t \leftarrow \text{learnable scalar}$$

The logit scale applied to similarities is then:

$$\text{logit scale} = \frac{1}{\tau} = \exp(-t)$$

### Initialization and Training Dynamics

- $t$ is initialized to $\log(1/0.07) \approx 2.66$, giving $\tau = 0.07$ at the start (matching the common fixed value)
- During training, the model **learns to increase** $\exp(-t)$ — sharpening the similarity distribution as representations improve
- CLIP **clamps** $\exp(t) \leq 100$ (equivalently, logit scale $\leq 100$) to prevent numerical instability

### Why Learnable?

| Fixed $\tau = 0.07$ | Learnable $\tau = \exp(t)$ |
|---------------------|---------------------------|
| Same sharpness throughout training | Adapts as representations improve |
| Early training may be too sharp | Starts soft, gradually sharpens |
| Requires manual tuning per dataset | Self-tunes to data distribution |

Early in training, embeddings are random and similarities are noisy — a softer distribution prevents overconfident wrong predictions. As the model learns meaningful structure, increasing sharpness (lower $\tau$) forces finer discrimination between similar negatives.

## SigLIP: Sigmoid Alternative to InfoNCE

While InfoNCE uses a **softmax over the entire batch** (treating classification as a multi-class problem), **SigLIP** (Sigmoid Loss for Language-Image Pre-training) treats each image-text pair as an independent **binary classification** problem.

### SigLIP Loss

$$\mathcal{L}_{\text{SigLIP}} = -\frac{1}{N^2}\sum_{i,j} \log \sigma\!\left((-1)^{\mathbb{1}[i \neq j]} \cdot z_{ij}\right)$$

where:
- $z_{ij} = s_{ij}/\tau + b$ — scaled similarity plus learnable bias $b$
- $\sigma(\cdot)$ — sigmoid function
- $(-1)^{\mathbb{1}[i \neq j]}$ — $+1$ for positive pairs $(i,i)$, $-1$ for negative pairs $(i,j), i \neq j$

Positive pairs are pushed toward $\sigma(z) \to 1$; negatives toward $\sigma(-z) \to 1$ (i.e., $z \to -\infty$).

### InfoNCE vs SigLIP

| Property | InfoNCE (Softmax) | SigLIP (Sigmoid) |
|----------|-------------------|------------------|
| **Normalization** | Softmax over batch (global) | Sigmoid per pair (local) |
| **Negatives required** | Must be in same batch | Any pairwise negatives |
| **Distributed training** | Requires all-gather of embeddings | No all-gather needed |
| **Scaling** | Degrades with batch fragmentation | Scales to arbitrary batch sizes |
| **Loss coupling** | All pairs compete in denominator | Each pair independent |

**Key advantage:** SigLIP eliminates the expensive all-gather operation in distributed training. Each GPU can compute its portion of the $N^2$ pairwise loss independently, making it significantly more scalable for large-scale pretraining.

## 3. Training Dynamics — Watch Learning Happen

```python
# Mini experiment: train two small encoders and watch the similarity matrix evolve

class TinyEncoder(nn.Module):
    def __init__(self, in_dim, out_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 64), nn.ReLU(),
            nn.Linear(64, out_dim)
        )
    def forward(self, x):
        return F.normalize(self.net(x), dim=-1)

# Create 8 distinct pairs (each pair shares a pattern)
N = 8
torch.manual_seed(42)
img_data = torch.randn(N, 32)
txt_data = torch.randn(N, 32)
# Make matching pairs more similar in hidden structure
shared_signal = torch.randn(N, 16)
img_data[:, :16] += shared_signal * 2
txt_data[:, :16] += shared_signal * 2

img_enc = TinyEncoder(32, 16)
txt_enc = TinyEncoder(32, 16)
optimizer = torch.optim.Adam(list(img_enc.parameters()) + list(txt_enc.parameters()), lr=1e-3)

# Train and capture snapshots
snapshots = []
snapshot_epochs = [0, 5, 20, 50, 100, 200]
losses = []

for epoch in range(201):
    img_emb = img_enc(img_data)
    txt_emb = txt_enc(txt_data)
    
    logits = img_emb @ txt_emb.T / 0.07
    labels = torch.arange(N)
    loss = (F.cross_entropy(logits, labels) + F.cross_entropy(logits.T, labels)) / 2
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    losses.append(loss.item())
    
    if epoch in snapshot_epochs:
        with torch.no_grad():
            sim = (img_enc(img_data) @ txt_enc(txt_data).T).numpy()
            snapshots.append((epoch, sim))

# Plot evolution
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Contrastive Learning: Similarity Matrix Evolution', 
             fontsize=16, fontweight='bold')

for ax, (epoch, sim) in zip(axes.flat, snapshots):
    im = ax.imshow(sim, cmap='RdBu_r', vmin=-1, vmax=1)
    ax.set_title(f'Epoch {epoch} (loss={losses[epoch]:.3f})', fontsize=12)
    ax.set_xlabel('Text')
    ax.set_ylabel('Image')
    
    diag_mean = np.mean(np.diag(sim))
    offdiag_mean = np.mean(sim[~np.eye(N, dtype=bool)])
    ax.text(0.02, 0.98, f'diag: {diag_mean:.2f}\noff: {offdiag_mean:.2f}',
            transform=ax.transAxes, va='top', fontsize=9,
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.tight_layout()
plt.savefig('../assets/training_evolution.png', dpi=150, bbox_inches='tight')
plt.show()

# Loss curve
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(losses, color='#9B59B6', linewidth=2)
ax.set_xlabel('Epoch')
ax.set_ylabel('InfoNCE Loss')
ax.set_title('Loss Curve', fontsize=14, fontweight='bold')
for epoch in snapshot_epochs:
    ax.axvline(x=epoch, color='gray', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Watch how the diagonal gets brighter (matching pairs) and off-diagonal gets darker!")
```

## Hard Negative Mining Strategies

Hard negatives are the primary driver of contrastive learning quality. Without them, the loss vanishes and gradients become negligible. Production systems employ several strategies to ensure a steady supply of challenging negatives:

### 1. In-Batch Negatives
Use all other samples in the current mini-batch as negatives for each anchor. This is **CLIP's approach** — with batch size $N$, each sample gets $N-1$ negatives "for free" without extra forward passes.

### 2. Cross-Batch Memory Bank
Store embeddings from recent batches in a fixed-size queue (MoCo-style). Sample negatives from this memory bank, decoupling the number of negatives from batch size. Enables large effective negative sets with modest GPU memory.

### 3. Hard Negative Mining
Explicitly compute pairwise similarities and select the **top-$k$ most confusing** non-matching pairs. These are negatives the model currently scores as highly similar — they produce the largest gradients.

### 4. Synthetic Negatives
Augment text captions with controlled perturbations to create semantically close but incorrect pairs:
- "a **red** car" → "a **blue** car"
- "a cat **sitting**" → "a cat **standing**"

These are guaranteed hard negatives because they share most tokens/context.

### Probability of Finding a Hard Negative

If each random sample has probability $p$ of being a hard negative, the probability of finding **at least one** hard negative in a batch of size $N$ is:

$$P(\text{hard neg}) = 1 - (1-p)^N$$

| Batch Size $N$ | $p = 0.01$ | $p = 0.05$ | $p = 0.10$ |
|----------------|------------|------------|------------|
| 32 | 0.27 | 0.80 | 0.96 |
| 256 | 0.92 | ≈ 1.00 | ≈ 1.00 |
| 32,768 | ≈ 1.00 | ≈ 1.00 | ≈ 1.00 |

This explains why CLIP's batch size of 32,768 was so effective — even rare hard negatives ($p = 0.01$) are virtually guaranteed in every batch.

## Key Takeaways

1. **InfoNCE loss** = softmax over similarities, matching pair as target
2. **Temperature** controls how peaked the probability distribution is (0.07 is common)
3. **Hard negatives** are critical — larger batches help find them
4. Training makes **diagonal bright, off-diagonal dark** in the similarity matrix
5. **Symmetric loss** (image→text AND text→image) works better
